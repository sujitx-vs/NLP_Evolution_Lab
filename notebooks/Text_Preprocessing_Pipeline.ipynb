# Text Preprocessing Pipeline using Python and NLTK

This notebook demonstrates basic NLP preprocessing steps.

```python
import re
import nltk
from nltk.tokenize import sent_tokenize
nltk.download("punkt_tab")
```

## Step 0: Read the Input Text

```python
file_path = "../datasets/txt_sample01.txt"

with open(file_path, "r", encoding="utf-8") as file:
    text = file.read()

print("=" * 60)
print("STEP 0 : ORIGINAL TEXT")
print("=" * 60)
print(text)
```

## Step 1: Whitespace Normalization
Normalize multiple whitespaces into one.

```python
normalized_text = re.sub(r"\s+", " ", text).strip()

print("=" * 60)
print("STEP 1 : NORMALIZED TEXT")
print("=" * 60)
print(normalized_text)
```

## Step 2: Naive Sentence Segmentation

```python
raw_sentences = normalized_text.split(".")
sentences = [s.strip() for s in raw_sentences if s.strip()]
print("=" * 60)
print("STEP 2 : SENTENCES")
print("=" * 60)
for i, sentence in enumerate(sentences, start=1):
    print(f"Sentence {i}: {sentence}")
```

## Step 3: Tokenization

```python
document_tokens=[]
print("="*60)
print("STEP 3 : TOKENS PER SENTENCE")
print("="*60)
for i,sentence in enumerate(sentences,1):
    tokens=sentence.split()
    document_tokens.append(tokens)
    print(f"\nSentence {i}")
    print(tokens)
```

## Step 4: Rule-Based Cleaning

```python
cleaned_document=[]
print("="*60)
print("STEP 4 : RULE-BASED CLEANING")
print("="*60)
for i,tokens in enumerate(document_tokens,1):
    cleaned_tokens=[t.strip(".,!?$") for t in tokens]
    cleaned_document.append(cleaned_tokens)
    print(f"\nSentence {i}")
    print(cleaned_tokens)
```

## Step 5: Regex-Based Cleaning

```python
regex_document=[]
print("="*60)
print("STEP 5 : REGEX-BASED CLEANING")
print("="*60)
for i,tokens in enumerate(document_tokens,1):
    regex_tokens=[re.sub(r"[^\w\s]","",t) for t in tokens]
    regex_document.append(regex_tokens)
    print(f"\nSentence {i}")
    print(regex_tokens)
```

## Step 6: NLTK Sentence Tokenizer

```python
nltk_sentences=sent_tokenize(text)
print("="*60)
print("STEP 6 : NLTK SENTENCE TOKENIZER")
print("="*60)
for i,s in enumerate(nltk_sentences,1):
    print(f"Sentence {i}: {s}")
```

## STEP 7 : NLTK WORD TOKENIZER

```python
from nltk.tokenize import word_tokenize

print("\n" + "=" * 60)
print("STEP 7 : NLTK WORD TOKENIZER")
print("=" * 60)

for i, sentence in enumerate(nltk_sentences, start=1):
    tokens = word_tokenize(sentence)

    print(f"\nSentence {i}")
    print(tokens)
```
