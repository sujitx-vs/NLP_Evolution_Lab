```python
from pypdf import PdfReader
```

```python
from pypdf import PdfReader

# Load your PDF file
reader = PdfReader(r"..\datasets\Sample01.pdf")

# Read text from all pages
for index, page in enumerate(reader.pages):
    text = page.extract_text()
    print(f"--- Page {index + 1} ---")
    print(text)
```

```python
len(reader.pages) # Return no of pages in the PDF.
```

```python
reader.metadata #metadata of the PDF file.
```

```python
print(reader.pages[0].extract_text()) # only first page text.
```

For text-based PDFs, PyPDF is one of the best lightweight and efficient libraries for extracting plain text. It is fast, has very few dependencies, and works well when the primary goal is to retrieve the textual content of a PDF.

PyPDF does not understand the semantic structure of tables. It simply extracts the text in reading order. Although the text contained inside a table may be extracted, the table's row-column structure is not preserved as a table object.

```python
import pypdfium2 as pdfium
text = "--".join(
    p.get_textpage().get_text_range() 
    for p in pdfium.PdfDocument(PATH)
)
print(text)
```

```python
# %pip install pdfplumber
```

```python
import pdfplumber
with pdfplumber.open(PATH) as pdf:
    first_page = pdf.pages[0]
    text = first_page.extract_text()
    table = first_page.extract_table()

print(text)
print(table)
```
